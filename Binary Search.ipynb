```python
class Solution:
    def firstSearch(self, arr, k):
        low = 0
        high = len(arr) - 1
        ans = -1

        while low <= high:
            mid = low + (high - low) // 2

            if arr[mid] == k:
                ans = mid        # Record the potential first occurrence
                high = mid - 1   # Keep searching left for an earlier match
            elif arr[mid] < k:
                low = mid + 1    # Search the right half
            else:
                high = mid - 1   # Search the left half

        return ans
```
